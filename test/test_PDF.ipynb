```python
import pymupdf
import re

def extract_pdf_text(pdf_path):
    doc = pymupdf.open(pdf_path)
    # 빈 리스트 pages 생성
    pages = []
    
    # 각 페이지에 대해:
    #     텍스트 추출
    #     텍스트가 비어있지 않으면 pages에 추가
    for page in doc:
        text = page.get_text()
        
        if text.strip():
            pages.append(text)
    
    # 문서 닫기
    doc.close()
    
    # pages가 비어있으면:
    #     ValueError 발생
    
    if not pages:
        raise ValueError("ValueError 발생") 
    
    # pages를 하나의 문자열로 합치기
    pages = "\n".join(pages)
    # _clean_text 호출 후 반환
    return _clean_text(pages)

def _clean_text(text):
    # 줄 단위로 쪼개기
    text = text.splitlines()
    # 빈 set 생성 (seen)
    seen = set()
    # 빈 리스트 생성 (result)
    result = []
    
    # 각 줄에 대해:
    #     앞뒤 공백 제거
    #     비어있지 않고 seen에 없으면:
    #         seen에 추가
    #         result에 추가
    for line in text:
        stripped = line.strip()
        
        if stripped and stripped not in seen:
            seen.add(stripped)
            result.append(stripped)
    
    # result를 줄바꿈으로 합치기
    # 연속 빈 줄 3개 이상 → 2개로 줄이기
    # 반환
    result = "\n".join(result)
    text = re.sub(r"(\n\s*){3,}","\n\n",result)
    return text

text = extract_pdf_text(r"C:\Users\dbtjd\Downloads\[WEB] 1. 웹프로그래밍개요.pdf")
print(text[:500])
print(f"\n총 길이: {len(text)}자")
```
